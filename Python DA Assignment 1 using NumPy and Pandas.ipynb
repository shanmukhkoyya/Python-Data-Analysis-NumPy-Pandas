```python
# Week 1: Daily average temperatures in Celsius
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
# Inspect the properties of the Week 1 temperature array

print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# Convert Celsius to Fahrenheit
temperatures_w1_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_w1_f)
```

```python
# Find maximum, minimum, and mean temperatures

print("Maximum temperature:", np.max(temperatures_w1), "°C")
print("Minimum temperature:", np.min(temperatures_w1), "°C")
print("Mean temperature:", np.mean(temperatures_w1), "°C")
```

```python
# First three days of the week
print(temperatures_w1[:3])
```

```python
# Weekend temperatures: last two days
print(temperatures_w1[-2:])
```

```python
# Middle three days of the week
print(temperatures_w1[2:5])
```

```python
# Create a 2D array for two weeks of temperatures

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

```python
# Inspect the 2D temperature array

print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)
```

```python
# Extract temperatures for each week

print("Week 1 temperatures:", temperatures[0])
print("Week 2 temperatures:", temperatures[1])
```

```python
# Extract weekend temperatures for both weeks

print("Week 1 weekend:", temperatures[0, -2:])
print("Week 2 weekend:", temperatures[1, -2:])
```

```python
# Create a Pandas Series with custom index labels

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
# 1. Access the 1st rank student using integer position
print("1st rank mark:", marks.iloc[0])

# 2. Retrieve the top 3 ranks using index labels
print("\nTop 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# 3. Retrieve the 3rd rank student's mark using iloc
print("\n3rd rank mark:", marks.iloc[2])

# 4. Filter ranks where marks are greater than 90
print("\nRanks with marks greater than 90:")
print(marks[marks > 90])
```

```python
# 1. Change the 1st rank student's mark to 100
marks.loc['Rank1'] = 100

print("After updating Rank1:")
print(marks)

# 2. Remove the last rank (Rank5)
marks = marks.drop('Rank5')

print("\nAfter removing Rank5:")
print(marks)

# 3. Calculate CGPA by dividing each mark by 10
cgpa = marks / 10

print("\nCGPA:")
print(cgpa)
```

```python
# Import pandas if you have not already imported it
import pandas as pd

# Create the transactions DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

# Display the DataFrame
print(transactions)
```

```python
# Display the complete DataFrame
display(transactions)

# Display the first 5 rows
print("First 5 rows:")
display(transactions.head())

# Display the last 5 rows
print("Last 5 rows:")
display(transactions.tail())

# Display the shape
print("Shape:", transactions.shape)

# Display column names
print("Column names:", transactions.columns.tolist())

# Display data types
print("Data types:")
print(transactions.dtypes)
```

```python
transactions.groupby('Region')['Amount'].mean()
```

```python
# Update the Amount to 165 for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

# Display the updated transaction
display(transactions[transactions['TransactionID'] == 102])
```

```python
# Calculate a 10% discount for each transaction
transactions['Discount'] = transactions['Amount'] * 0.10

# Display the DataFrame
display(transactions)
```

```python
# Remove TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Display the updated DataFrame
display(transactions)
```

```python
# Delete the Discount column
transactions = transactions.drop(columns=['Discount'])

# Display the final DataFrame
display(transactions)
```
